```python
# ทดสอบการใช้ load_dotenv() เพื่อโหลดค่า GROQ_API_KEY จากไฟล์ .env
from dotenv import load_dotenv
import os
from groq import Groq
import os

load_dotenv()
groq_api_key = os.getenv('GROQ_API_KEY')
```

```python
groq_api_key = os.environ.get('GROQ_API_KEY')
client = Groq(api_key=groq_api_key)
messages = [
    {"role": "system", "content": "You are a knowledgeable historian and AI assistant specializing in historical events."},
    {"role": "user", "content": "What caused World War I?"}
 ]
response = client.chat.completions.create(
    messages=messages,
    model="llama-3.3-70b-versatile",
    temperature=0.3,
    stream=False
 )
print('AI Response:', response.choices[0].message.content)
```
